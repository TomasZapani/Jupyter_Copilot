```python
from IPython.display import HTML

HTML("""
<style>
#llm-copilot {
    position: fixed;
    top: 80px;
    right: 30px;
    width: 350px;
    background: #111;
    color: #fff;
    padding: 10px;
    border-radius: 8px;
    border: 2px solid #444;
    z-index: 999999;
    font-family: Arial;
    resize: both;
    overflow: auto;
}
#llm-header {
    cursor: move;
    background: #222;
    padding: 5px;
    border-radius: 6px 6px 0 0;
}
#llm-minimize {
    float: right;
    cursor: pointer;
    margin-left: 10px;
    font-weight: bold;
}
</style>

<div id='llm-copilot'>
    <div id='llm-header'>
        💬 LLM Copilot
        <span id='llm-minimize'>−</span>
    </div>

    <textarea id='prompt' style='width:100%; height:120px;'
              onkeydown="event.stopPropagation();"></textarea>

    <button onclick='sendPrompt()' style='margin-top:10px; width:100%;'>
        Enviar
    </button>

    <pre id='result' style='white-space:pre-wrap; margin-top:10px;'></pre>
</div>

<script>
// --- Evitar que Jupyter capture teclas ---
document.getElementById("llm-copilot").addEventListener("keydown", e => {
    e.stopPropagation();
});

// --- Enviar al backend ---
function sendPrompt() {
    let msg = document.getElementById("prompt").value;
    fetch("http://127.0.0.1:8000/generate", {
        method:"POST",
        headers:{ "Content-Type":"application/json" },
        body:JSON.stringify({message: msg})
    })
    .then(r => r.json())
    .then(d => document.getElementById("result").innerText = d.response);
}

// --- Minimizar ---
document.getElementById("llm-minimize").onclick = () => {
    let box = document.getElementById("llm-copilot");
    if (box.style.height === "30px") {
        box.style.height = "auto";
        box.style.padding = "10px";
    } else {
        box.style.height = "30px";
        box.style.padding = "5px";
    }
};

// --- Hacerlo arrastrable ---
(function() {
    let box = document.getElementById("llm-copilot");
    let header = document.getElementById("llm-header");
    let offsetX = 0, offsetY = 0, startX = 0, startY = 0;

    header.onmousedown = e => {
        startX = e.clientX;
        startY = e.clientY;
        document.onmousemove = drag;
        document.onmouseup = stopDrag;
    };

    function drag(e) {
        offsetX = e.clientX - startX;
        offsetY = e.clientY - startY;
        startX = e.clientX;
        startY = e.clientY;
        box.style.top = (box.offsetTop + offsetY) + "px";
        box.style.left = (box.offsetLeft + offsetX) + "px";
    }

    function stopDrag() {
        document.onmousemove = null;
        document.onmouseup = null;
    }
})();
</script>
""")
```
